### Observations

Smaller ε -> reccurence rate shrinks
Larger ε -> reccurence rate grows

It is only natural that such values of ε will lead to such results. What is important is determining how to choose an ε s.t. results aren't forced in either direction : we want the recurrence plot to capture enough information, while filtering out the noise.

```python
# Randomly chosen values, not sure if they are good/bad
m = 4
t = 2

shape = (len(x_t - (m - 1) * t), len(x_t - (m - 1) * t))
R1, R2, R3, R4 = [np.zeros(shape, dtype=bool) for _ in range(4)]

windows = np.lib.stride_tricks.sliding_window_view(x_t, (m - 1) * t + 1)
vectors = windows[:, ::t] # keep slices with a delay of t

shape = (len(vectors), len(vectors))
R1, R2, R3, R4 = [np.zeros(shape, dtype=bool) for _ in range(4)]
#windows
```

```python
#vectors
```

```python
for i in range(len(vectors)):
    for j in range(len(vectors)):

        # Use Euc Distance
        Δ = (vectors[i] - vectors[j]) ** 2
        Δ = np.sqrt(np.sum(Δ))

        R1[i,j] = Δ <= ε1
        R2[i,j] = Δ <= ε2
        R3[i,j] = Δ <= ε3
        R4[i,j] = Δ <= ε4
```

### Randomness

```python
t = np.arange(0, 100)
x_t = np.sin(0.04 * np.pi * t)

rng = np.random.default_rng(seed=42)
noise_strength = 0.25
ε = 0.25

# Noise
added_noise = rng.normal(loc=0.0, scale=noise_strength, size=len(t))
x_noisy = x_t + added_noise
x_random = rng.normal(loc=0.0, scale=noise_strength, size=len(t))
```

```python
n = len(t)
R = np.zeros((n, n), dtype=bool)
R_noisy = np.zeros((n, n), dtype=bool)
R_random = np.zeros((n, n), dtype=bool)

for i in range(len(t)):
   for j in range(len(t)):

       Δ = abs(x_t[i] - x_t[j])
       Δ_noisy = abs(x_noisy[i] - x_noisy[j])
       Δ_random = abs(x_random[i] - x_random[j])

       R[i, j] = Δ <= ε
       R_noisy[i, j] = Δ_noisy <= ε
       R_random[i, j] = Δ_random <= ε
```

```python
fig, ax = plt.subplots(3, 1, figsize=(10, 8), sharex=True)

ax[0].plot(t, x_t)
ax[0].set_title("Sin Wave")
ax[0].set_ylabel("x(t)")
ax[0].grid()

ax[1].plot(t, x_noisy)
ax[1].set_title("Sin wave w/ added noise")
ax[1].set_ylabel("x(t)")
ax[1].grid()

ax[2].plot(t, x_random)
ax[2].set_title("Independent random noise")
ax[2].set_xlabel("t")
ax[2].set_ylabel("x(t)")
ax[2].grid()

plt.tight_layout()
plt.show()
```

### Reading RPs

**What do long diagonal lines indicate?**

A stretch of the signal stays close to another stretch for many steps, so the dynamics repeat in a predictable way. On the sine wave that shows up as long, evenly spaced diagonals parallel to the main diagonal, one for each multiple of the period.

**What might short diagonal lines indicate?**

States are similar only briefly, then drift apart. That is short-lived determinism: chaos, a changing regime, or a longer diagonal that has already been broken

**What does a collection of isolated recurrence points suggest?**

Those pairs of times happen to fall inside ε, but their neighbours do not. The closeness is accidental, which is what uncorrelated noise looks like: a scatter of single points + the main diagonal

**How does adding noise affect the diagonal structures?**

Noise pushes samples outside ε at random times, so long diagonals split into shorter pieces and isolated points. With moderate noise the periodic pattern is still visible; with enough noise it disappears into the random-scatter plot

**Why does a periodic signal produce a visibly different structure from independent noise?**

If `x(t) ≈ x(t + kT)`, the samples just after those times are close as well, so recurrences line up into diagonals. Independent noise has no such link: `|x_i − x_j| ≤ ε` says nothing about `|x_{i+1} − x_{j+1}|`, so the points stay scattered.

### Joint recurrences

Three binary series, w/ length \(N = 100\). Equality determines recurrence :

\[
R^X_{ij} = 1\{X_i = X_j\},
\]

Same for \(Y\) and \(Z\).


### Individual RPs

1. \(X, Y, Z\) i.i.d. Bernoulli\((1/2)\).
2. \(X, Y\) i.i.d. Bernoulli\((1/2)\), and \(Z = X \oplus Y\).
3. \(X = Y \sim\) Bernoulli\((1/2)\), with \(Z \sim\) Bernoulli\((1/2)\) independent of \(X\).

**Do these show the dependence?**

Only the copy in model 3. There \(Y\) is \(X\), so \(R^X\) and \(R^Y\) are the same matrix. Models 1 and 2 cannot be told apart this way: every margin is a fair coin toss, and a single recurrence plot never compares one series with another. The XOR link does not appear in \(R^X\), \(R^Y\), or \(R^Z\) separately.


```python
N = 100
rng = np.random.default_rng(42)

def fair_bits(n):
    return rng.integers(0, 2, size=n)

X1, Y1, Z1 = fair_bits(N), fair_bits(N), fair_bits(N)

X2, Y2 = fair_bits(N), fair_bits(N)
Z2 = X2 ^ Y2

X3 = fair_bits(N)
Y3 = X3.copy()
Z3 = fair_bits(N)

models = {
    "Independent": (X1, Y1, Z1),
    "XOR": (X2, Y2, Z2),
    "X = Y": (X3, Y3, Z3),
}

def R_of(x):
    return x[:, None] == x[None, :]

fig, ax = plt.subplots(3, 3, figsize=(9.5, 9), squeeze=False)
for r, (name, (X, Y, Z)) in enumerate(models.items()):
    for c, (label, series) in enumerate(zip(["$R^X$", "$R^Y$", "$R^Z$"], (X, Y, Z))):
        ax[r, c].imshow(
            R_of(series),
            cmap="copper",
            origin="lower",
            interpolation="nearest",
        )
        ax[r, c].set_title(f"{name}: {label}")
        ax[r, c].set_xlabel("j")
        ax[r, c].set_ylabel("i")

fig.suptitle("Individual recurrence plots")
fig.tight_layout()
plt.show()
```

### Pairwise joint RPs

\[
J^{XY}_{ij} = R^X_{ij}\, R^Y_{ij},
\]

and likewise for \(J^{XZ}\) and \(J^{YZ}\). A point appears iff both series recur.

The joint plots are thinner than the individual ones, because both matches are required.

- **Independent** Each joint plot is the overlap of two unrelated patterns.
- **XOR** The three pairwise plots are the same matrix. A match on any two coordinates forces a match on the third, so the third indicator removes nothing.
- **\(X = Y\)** \(J^{XY} = R^X\), because the condition on \(Y\) repeats the condition on \(X\). \(J^{XZ}\) and \(J^{YZ}\) are the same thinner plot.


```python
pairwise = []
for RX, RY, RZ in individual:
    pairwise.append((RX & RY, RX & RZ, RY & RZ))

show_grid(
    pairwise,
    list(models),
    ["$J^{XY}$", "$J^{XZ}$", "$J^{YZ}$"],
    "Pairwise joint RPs"
)

print("XOR: JXY equals JXZ?", np.array_equal(pairwise[1][0], pairwise[1][1]))
print("XOR: JXY equals JYZ?", np.array_equal(pairwise[1][0], pairwise[1][2]))
print("X = Y: JXY equals RX?", np.array_equal(pairwise[2][0], individual[2][0]))
print("X = Y: JXZ equals JYZ?", np.array_equal(pairwise[2][1], pairwise[2][2]))
```

### Triple joint recurrence plot

\[
J^{XYZ}_{ij} = R^X_{ij}\, R^Y_{ij}\, R^Z_{ij}.
\]

**Why, in the XOR model, any two recurrences imply the third.** Write \(Z_i = X_i \oplus Y_i\) and \(Z_j = X_j \oplus Y_j\). XOR is its own inverse: \(A \oplus B = C\) gives \(A = B \oplus C\).

- \(X_i = X_j\) and \(Y_i = Y_j\) give \(Z_i = X_i \oplus Y_i = X_j \oplus Y_j = Z_j\).
- \(X_i = X_j\) and \(Z_i = Z_j\) give \(Y_i = X_i \oplus Z_i = X_j \oplus Z_j = Y_j\).
- \(Y_i = Y_j\) and \(Z_i = Z_j\) give \(X_i = Y_i \oplus Z_i = Y_j \oplus Z_j = X_j\).

So on every XOR sample, \(J^{XY} = J^{XZ} = J^{YZ} = J^{XYZ}\).


```python
triple = [(RX & RY & RZ,) for RX, RY, RZ in individual]

show_grid(
    triple,
    list(models),
    ["$J^{XYZ}$"],
    "Triple joint recurrence plots"
)

Jxy, Jxz, Jyz = pairwise[1]
Jxyz = triple[1][0]
print("XOR: JXY equals JXYZ?", np.array_equal(Jxy, Jxyz))
print("XOR: JXZ equals JXYZ?", np.array_equal(Jxz, Jxyz))
print("XOR: JYZ equals JXYZ?", np.array_equal(Jyz, Jxyz))
```

### Individual, pairwise, and triple recurrence rates

Same off-diagonal pairs \(i < j\) for every matrix:

\[
\hat q = \frac{2}{N(N-1)} \sum_{i<j} R_{ij}.
\]

**What separates the models.** Every margin is a fair bit, so all three individual rates sit near \(1/2\).

The independent model and the XOR model also share the pairwise rates, near \(1/4\). They separate at the triple rate: near \(1/8\) when the three series are independent, near \(1/4\) under XOR. Individual and pairwise rates are not enough to tell those two apart.

\(X = Y\) is already different in the pairs. \(\hat q_{XY}\) is near \(1/2\), while \(\hat q_{XZ}\) and \(\hat q_{YZ}\) stay near \(1/4\), and the triple rate sits near \(1/4\).


```python
def off_diagonal_rate(R):
    n = R.shape[0]
    upper = np.triu_indices(n, k=1) # entries above diagonal
    return float(R[upper].mean())

def rate_row(RX, RY, RZ):
    Jxy = RX & RY
    Jxz = RX & RZ
    Jyz = RY & RZ
    Jxyz = Jxy & RZ
    return {
        "qX": off_diagonal_rate(RX),
        "qY": off_diagonal_rate(RY),
        "qZ": off_diagonal_rate(RZ),
        "qXY": off_diagonal_rate(Jxy),
        "qXZ": off_diagonal_rate(Jxz),
        "qYZ": off_diagonal_rate(Jyz),
        "qXYZ": off_diagonal_rate(Jxyz)
    }

rate_keys = ["qX", "qY", "qZ", "qXY", "qXZ", "qYZ", "qXYZ"]
rates = {
    name: rate_row(*mats)
    for name, mats in zip(models, individual)
}

print(f"{'model':<14}" + "".join(f"{key:>8}" for key in rate_keys))
for name, row in rates.items():
    print(f"{name:<14}" + "".join(f"{row[key]:8.3f}" for key in rate_keys))
```

### Population recurrence probabilities

Take two independent draws, \((X, Y, Z)\) and \((X', Y', Z')\), from the same model.

\[
q_X = P(X = X'), \qquad
q_{XY} = P(X = X',\, Y = Y'), \qquad
q_{XYZ} = P(X = X',\, Y = Y',\, Z = Z').
\]

**Independent Bernoulli\((1/2)\).** A fair bit matches an independent copy with probability \(1/2\), so \(q_X = q_Y = q_Z = 1/2\). The three coordinates are independent, and the copy is independent of the original, so the match events multiply:

\[
q_{XY} = q_{XZ} = q_{YZ} = \tfrac14, \qquad q_{XYZ} = \tfrac18.
\]

**XOR.** \(Z = X \oplus Y\) is still Bernoulli\((1/2)\), and Step 2 showed that each pair \((X, Y)\), \((X, Z)\), \((Y, Z)\) is independent. The second draw is an independent copy, so the single-variable and pairwise values stay \(1/2\) and \(1/4\). The triple does not. \(X = X'\) and \(Y = Y'\) already force \(Z = Z'\), hence

\[
q_{XYZ} = P(X = X',\, Y = Y') = q_{XY} = \tfrac14.
\]

**\(X = Y\), \(Z\) independent.** Again \(q_X = q_Y = q_Z = 1/2\). Because \(Y = X\) and \(Y' = X'\), the event \(Y = Y'\) is the same event as \(X = X'\), so \(q_{XY} = 1/2\). \(Z\) is an independent fair bit, so \(q_{XZ} = q_{YZ} = 1/4\). Once \(X\) matches, the condition on \(Y\) adds nothing:

\[
q_{XYZ} = P(X = X',\, Z = Z') = \tfrac14.
\]

| Model | \(q_X, q_Y, q_Z\) | \(q_{XY}, q_{XZ}, q_{YZ}\) | \(q_{XYZ}\) |
| --- | --- | --- | --- |
| Independent | \(1/2,\, 1/2,\, 1/2\) | \(1/4,\, 1/4,\, 1/4\) | \(1/8\) |
| XOR | \(1/2,\, 1/2,\, 1/2\) | \(1/4,\, 1/4,\, 1/4\) | \(1/4\) |
| \(X = Y\) | \(1/2,\, 1/2,\, 1/2\) | \(1/2,\, 1/4,\, 1/4\) | \(1/4\) |

The rates here are the sample versions of these probabilities. At \(N = 100\) they fall near the ones in the table, converging towards that value as \(N\) grows. As any sample will be finite, it won't hit the population values exactly.


```python
theory = {
    "Independent": {
        "qX": 1/2, "qY": 1/2, "qZ": 1/2,
        "qXY": 1/4, "qXZ": 1/4, "qYZ": 1/4,
        "qXYZ": 1/8
    },
    "XOR": {
        "qX": 1/2, "qY": 1/2, "qZ": 1/2,
        "qXY": 1/4, "qXZ": 1/4, "qYZ": 1/4,
        "qXYZ": 1/4
    },
    "X = Y": {
        "qX": 1/2, "qY": 1/2, "qZ": 1/2,
        "qXY": 1/2, "qXZ": 1/4, "qYZ": 1/4,
        "qXYZ": 1/4,
    }
}

print(f"{'model':<14}" + "".join(f"{key:>8}" for key in rate_keys))
for name in models:
    sample = rates[name]
    exact = theory[name]
    print(f"{name:<14}" + "".join(f"{sample[key]:8.3f}" for key in rate_keys) + "   sample")
    print(f"{'':<14}" + "".join(f"{exact[key]:8.3f}" for key in rate_keys) + "   theory")
```

### A measure of higher-order dependence

\[
D_3 = q_{XYZ} - q_X q_Y q_Z.
\]

| Model | \(D_3\) |
| --- | --- |
| Three independent Bernoulli\((1/2)\) variables | \(0\) |
| \(Z = X \oplus Y\) | \(1/8\) |
| \(X = Y\), \(Z\) independent | \(1/8\) |

Both nonzero values are \(1/4 - (1/2)^3 = 1/8\). The independent model gives \(1/8 - 1/8 = 0\).

**Mutual independence implies \(D_3 = 0\).** If \(X, Y, Z\) are mutually independent, two independent copies have six mutually independent coordinates, and

\begin{align*}
q_{XYZ}
&= \sum_{x,y,z} P(X=x, Y=y, Z=z)\, P(X'=x, Y'=y, Z'=z) \\
&= \sum_{x,y,z} P(X=x)^2\, P(Y=y)^2\, P(Z=z)^2 \\
&= q_X\, q_Y\, q_Z.
\end{align*}

**The converse is false.** \(D_3 = 0\) does not imply mutual independence. Take the joint distribution

| | \(Y = 0\) | \(Y = 1\) |
| --- | --- | --- |
| \(X = 0\) | \(5/12\) | \(1/3\) |
| \(X = 1\) | \(1/4\) | \(0\) |

and let \(Z \sim\) Bernoulli\((1/2)\) be independent of \((X, Y)\). Then \(P(X=1) = 1/4\) and \(P(Y=1) = 1/3\), but \(P(X=1, Y=1) = 0 \neq 1/12\), so \(X\) and \(Y\) are dependent.

The recurrence probabilities still factor. \(q_X = (1/4)^2 + (3/4)^2 = 5/8\) and \(q_Y = (1/3)^2 + (2/3)^2 = 5/9\), while

\[
q_{XY} = (5/12)^2 + (1/3)^2 + (1/4)^2 = 25/72 = q_X q_Y.
\]

Independence of \(Z\) gives \(q_{XYZ} = q_{XY}\, q_Z = q_X q_Y q_Z\), so \(D_3 = 0\).

**Same score, different dependence.** XOR and \(X = Y\) both have \(D_3 = 1/8\).

- Under XOR, every pair has the independent-model rate \(q_{XY} = q_X q_Y = 1/4\). The dependence is visible only in the triple, where \(q_{XYZ} = 1/4\) rather than \(1/8\).
- Under \(X = Y\), \(Z\) is independent of the pair. \(D_3\) is nonzero because \(q_{XY} = 1/2 > q_X q_Y\), which lifts \(q_{XYZ}\) with it. There is no three-way interaction.

\(D_3\) records that the triple recurrence does not split into the product of the three marginal recurrences. It does not say whether that failure comes from a three-way link or from a pairwise link.


```python
print(f"{'model':<14}{'D3 theory':>12}{'D3 sample':>12}")
for name in models:
    exact = theory[name]
    sample = rates[name]
    D3 = exact["qXYZ"] - exact["qX"] * exact["qY"] * exact["qZ"]
    D3_hat = sample["qXYZ"] - sample["qX"] * sample["qY"] * sample["qZ"]
    print(f"{name:<14}{D3:12.3f}{D3_hat:12.3f}")

# Converse fails: dependent (X, Y), independent Z, but D3 = 0.
joint_xy = np.array([
    [5/12, 1/3],
    [1/4,  0.0],
])
px = joint_xy.sum(axis=1)
py = joint_xy.sum(axis=0)
qX = np.sum(px**2)
qY = np.sum(py**2)
qXY = np.sum(joint_xy**2)
qZ = 1/2
qXYZ = qXY * qZ
print()
print("counterexample P(X=1, Y=1) =", joint_xy[1, 1], " product =", px[1] * py[1])
print("counterexample D3 =", qXYZ - qX * qY * qZ)
```
